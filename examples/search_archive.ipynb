# Searching archive for SPEAR data

This notebook serves as an example for how to search for SPEAR datasets available in /archive before submitting the request form, in order to verify that the datasets you are interested in are available.

```python
import intake
import intake_esm
catalog_path = "/work/a3r/catalogs/spear_med_archive/catalog.json"
```

The first step is to load the catalog:

```python
cat = intake.open_esm_datastore(catalog_path)
```

Firstly, we should check what metadata fields and values are avaiable. To get a full list of columns in the catalog,

```python
cat.df.columns
```

and we can get the unique options for each column,

```python
def show_unique(catalog, column):
    print(set(catalog.df[column]))

show_unique(cat, "variable_id")
show_unique(cat, "frequency")
show_unique(cat, "realm")
```

To search for your desired data:

```python
subcat = cat.search(
    variable_id='t_ref', 
    frequency='daily', 
    realm='atmos_daily',
    experiment_id='SPEAR_c192_o1_Hist_AllForc_IC1921_K50'
)
subcat
```

Double check that all time_ranges are available for your desired data:

```python
show_unique(subcat, 'time_range')
```

Data for the historical experiment_id should have time_ranges from 1921 to 2010. If chunk_freq is 10 yr for the given variable, this would be 9 individual netCDF files, so for 30 members we would expect 270 unique entries in 'path'
