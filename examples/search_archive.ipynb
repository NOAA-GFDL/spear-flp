# Searching archive for SPEAR data

This notebook serves as an example for how to search for SPEAR datasets available in /archive before submitting the request form, in order to verify that the datasets you are interested in are available.

```python
import intake
import intake_esm
catalog_path = "/work/a3r/catalogs/spear_med_archive/catalog.json"
```

The first step is to load the catalog:

```python
cat = intake.open_esm_datastore(catalog_path)
```

Firstly, we should check what metadata fields and values are avaiable. To get a full list of columns in the catalog,

```python
cat.df
```

and we can get the unique options for each column,

```python
cat.df.realm.unique()
```

```python
cat.df.frequency.unique()
```

Let's now search for the variable we want:

```python
#Examples to just search for what we want from the catalog
exp_id = ['SPEAR_c192_o1_Hist_AllForc_IC1921_K50']
realm = 'atmos'
frequency = "monthly"
```

To search for your desired data:

```python
subcat = cat.search(
    frequency=frequency, 
    realm=realm,
    experiment_id=exp_id
)
subcat
```

Then we can list all of the variables avaiable for that realm, frequency, and experiment combination:

```python
subcat.df['standard_name'].unique()
```

To find the corresponding GFDL variable_id:

```python
subcat.search(
    standard_name="surface_temperature"
).df['variable_id'].unique()
```
